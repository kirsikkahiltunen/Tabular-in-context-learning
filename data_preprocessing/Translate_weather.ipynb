```python
import numpy as np
import pandas as pd
```

```python
#df = pd.read_csv("data/malmi.csv")
df = pd.read_csv("data/FI_kumpula_19-25.csv")
df.head()
```

```python
df = df.rename(columns={"Vuosi": "Year" , "Tunti (Paikallinen aika)":"Hour (Local time)", "Kuukausi": "Month", "Päivä": "Day", "Ilman lämpötila keskiarvo (°C)": "Average air temperature (°C)", "Pilvisyys (1/8)": "Cloudiness (1/8)", "Näkyvyys keskiarvo (m)": "Average visibility (m)", "Suhteellinen kosteus keskiarvo (%)": "Relative humidity average (%)", "Keskituulen nopeus keskiarvo (m/s)": "Average wind speed (m/s)", "Vallitseva sää": "Current weather", "Tuulen suunta keskiarvo (°)": "Average wind direction (°)"})
df.head()
```

```python
df.value_counts("Current weather")
```

```python
df.shape
```

```python
df = df.replace("Heikkoa lumisadetta", "Light snow")
df = df.replace("Heikkoa vesisadetta", "Light rain")
df = df.replace("Utua", "Mist")
df = df.replace("Heikkoja vesikuuroja", "Light rain showers")
df = df.replace("Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä", "Drizzle or snow grains during the preceding hour, but not at the time of observation")
df = df.replace("Heikkoa tai kohtalaista sadetta", "Light or moderate precipitation")
df = df.replace("Heikkoja lumikuuroja", "Light snow showers")
```

```python
df.head()
```

```python
df.to_csv("data/EN_kumpula_19-25.csv", index=False)
```
