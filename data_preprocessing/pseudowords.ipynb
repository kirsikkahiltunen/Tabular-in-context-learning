# Diagnoosit pseudo words

```python
import numpy as np
import pandas as pd
import random
import string
```

```python
df =  pd.read_csv("data/FI_diagnoosit.csv")
df_original = df
```

```python
def pseudo_words(length, list_length):
    words=[]
    for i in range(list_length):
        word = ''.join(random.choices(string.ascii_lowercase, k=length))
        word = word.capitalize()
        words.append(word)

    return (words)
```

```python
words=(pseudo_words(10, 41))
print(words)
```

```python
df = df.rename(columns={"aikatyyppi": words[0] , "kuukausi_nro": words[1], "vuosi": words[2], "ikaryhma": words[3], "sukupuoli": words[4], "etuus":words[5], "diagnoosi":words[6], "alkanut_kausi_lkm":words[7], "ka_paiva_lkm":words[8], "ka_maksettu/paiva":words[9]})
```

```python
df = df.replace("Vuosikertymä", words[10])
df = df.replace("Kuukausi", words[11])
df = df.replace("Nainen", words[12])
df = df.replace("Mies", words[13])
df = df.replace("F41 Muut ahdistuneisuushäiriöt", words[14])
df = df.replace("M54 Selkäsärky", words[15])
df = df.replace("F32 Masennustila", words[16])
df = df.replace("F43 Reaktiot vaikeaan stressiin ja sopeutumishäiriöt", words[17])
df = df.replace("F33 Toistuva masennus", words[18])
df = df.replace("F51 Ei-elimelliset unihäiriöt", words[19])
df = df.replace("M51 Muut selän nikamavälilevyjen sairaudet", words[20])
df = df.replace("M79 Muualla luokittamattomat muut pehmytkudossairaudet", words[21])
df = df.replace("S83 Polven alueen nivelten ja siteiden sijoiltaanmeno, nyrjähdys ja/tai venähdys", words[22])
df = df.replace("S82 Polven ja/tai säären murtuma", words[23])
df = df.replace("G56 Yläraajan yhden hermon sairaudet", words[24])
df = df.replace("M75 Hartianseudun pehmytkudossairaudet", words[25])
df = df.replace("M25 Muualla luokittamattomat muut nivelsairaudet", words[26])
df = df.replace("S52 Kyynärvarren murtuma", words[27])
df = df.replace("S92 Jalkaterän luiden murtumat", words[28])
df = df.replace("M77 Muut siteiden ja lihasjänteiden kiinnittymiskohtien sairaudet (entesopatiat)", words[29])
df = df.replace("S43 Hartianseudun (hartiakaaren) nivelten ja siteiden sijoiltaanmeno, nyrjähdys ja/tai venähdys", words[30])
df = df.replace("F31 Kaksisuuntainen mielialahäiriö", words[31])
df = df.replace("J06 Useissa tai määrittämättömissä ylähengitysteiden kohdissa esiintyvä akuutti infektio", words[32])
df = df.replace("S62 Ranteen tai käden murtuma", words[33])
df = df.replace("M23 Polven sisäiset viat", words[34])
df = df.replace("Sairauspäiväraha", words[35])
df = df.replace("Yhteensä", words[36])
df = df.replace("Osasairauspäiväraha", words[37])
df = df.replace("YEL-päiväraha", words[38])
df = df.replace("Tartuntatautipäiväraha", words[39])
df = df.replace("Luovutuspäiväraha", words[40])
```
