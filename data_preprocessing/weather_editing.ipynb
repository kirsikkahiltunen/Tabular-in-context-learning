```python
import numpy as np
import pandas as pd
```

```python
kumpula19 = pd.read_csv("data/Kumpula_2019.csv")
kumpula20 = pd.read_csv("data/Kumpula_2020.csv")
kumpula21 = pd.read_csv("data/Kumpula_2021.csv")
kumpula22 = pd.read_csv("data/Kumpula_2022.csv")
kumpula23 = pd.read_csv("data/Kumpula_2023.csv")
kumpula24 = pd.read_csv("data/Kumpula_2024.csv")
kumpula25 = pd.read_csv("data/Kumpula_2025.csv")
kumpula19 = kumpula19[kumpula19["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
kumpula20 = kumpula20[kumpula20["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
kumpula21 = kumpula21[kumpula21["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
kumpula22 = kumpula22[kumpula22["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
kumpula23 = kumpula23[kumpula23["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
kumpula24 = kumpula24[kumpula24["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
kumpula25 = kumpula25[kumpula25["Vallitseva sää"].isin(['Heikkoa lumisadetta', 'Heikkoja vesikuuroja', 'Utua', 'Heikkoa vesisadetta', 'Tihkusadetta tai lumijyväsiä edellisen tunnin aikana, mutta ei enää havaintohetkellä', 'Heikkoa tai kohtalaista sadetta', 'Heikkoja lumikuuroja'])]
```

```python
kumpula_19_25 = pd.concat([kumpula19, kumpula20, kumpula21, kumpula22, kumpula23, kumpula24, kumpula25])
kumpula_19_25.head()
```

```python
kumpula_19_25.shape
```

```python
kumpula_19_25["Pilvisyys [1/8]"] = kumpula_19_25["Pilvisyys [1/8]"].str[-4]
kumpula_19_25.columns = kumpula_19_25.columns.str.replace("[", "(")
kumpula_19_25.columns = kumpula_19_25.columns.str.replace("]", ")")
```

```python
#kumpula_19_25 = pd.read_csv("data/kumpula_19-25.csv")
```

```python
kumpula_19_25 = kumpula_19_25.sample(frac=1).reset_index(drop=True)
kumpula_19_25.head()
```

```python
kumpula_19_25["Tunti (Paikallinen aika)"] = kumpula_19_25["Aika (Paikallinen aika)"].str[0:2]
kumpula_19_25 = kumpula_19_25.drop(columns=["Havaintoasema", "Aika (Paikallinen aika)"])
```
