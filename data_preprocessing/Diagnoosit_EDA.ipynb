```python
df = df.drop(columns=["vuosikuukausi", "kunta_nro", "kunta_nimi", "sairauspaaryhma"])
```

```python
df.shape
```

```python
df = df[df["diagnoosi"] != "Yhteensä"]
df = df[df["diagnoosi"] != "Tieto puuttuu"]
df = df[df["saaja_lkm"] > 0]
df = df[df["maksettu_eur"] > 0]
```

```python
df = df[df["korvattu_paiva_lkm"] > 0]
```

```python
df = df.sample(frac=1).reset_index(drop=True)
```

```python
print((df["saaja_lkm"] == 0).sum())
print((df["maksettu_eur"] == 0).sum())
```

```python
print(df["saaja_lkm"].isna().sum())
print(df["korvattu_paiva_lkm"].isna().sum())
print(df["maksettu_eur"].isna().sum())
```

```python
print(df["saaja_lkm"].min())
print(df["maksettu_eur"].min())
print(df["korvattu_paiva_lkm"].min())
print(df["saaja_lkm"].max())
print(df["maksettu_eur"].max())
print(df["korvattu_paiva_lkm"].max())
```

```python
df["ka_paiva_lkm"] = df["korvattu_paiva_lkm"].div(df["saaja_lkm"])
df["ka_maksettu/paiva"] = df["maksettu_eur"].div(df["korvattu_paiva_lkm"])
df = df.round({"ka_paiva_lkm": 2, "ka_maksettu/paiva":2})
```

```python
df = df.drop(columns=["saaja_lkm", "maksettu_eur", "korvattu_paiva_lkm"])
```
