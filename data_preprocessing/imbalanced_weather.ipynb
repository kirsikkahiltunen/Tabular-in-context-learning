# Kumpula weather data imbalanced

```python
import numpy as np
import pandas as pd
```

```python
pd.set_option("display.max_rows", None)
```

```python
kumpula19 = pd.read_csv("data/Kumpula_2019.csv")
kumpula20 = pd.read_csv("data/Kumpula_2020.csv")
kumpula21 = pd.read_csv("data/Kumpula_2021.csv")
kumpula22 = pd.read_csv("data/Kumpula_2022.csv")
kumpula23 = pd.read_csv("data/Kumpula_2023.csv")
kumpula24 = pd.read_csv("data/Kumpula_2024.csv")
kumpula25 = pd.read_csv("data/Kumpula_2025.csv")
```

```python
df = pd.concat([kumpula19, kumpula20, kumpula21, kumpula22, kumpula23, kumpula24, kumpula25])
df.head()
```

```python
df.shape
```

```python
df["Pilvisyys [1/8]"] = df["Pilvisyys [1/8]"].str[-4]
```

```python
df = df.replace("-", None)
df.columns = df.columns.str.replace("[", "(")
df.columns = df.columns.str.replace("]", ")")
df.columns = df.columns.str.replace("<", " ")
df.columns = df.columns.str.replace(">", " ")
```

```python
df["Tunti (Paikallinen aika)"] = df["Aika (Paikallinen aika)"].str[0:2]
df = df.drop(columns=["Havaintoasema", "Aika (Paikallinen aika)"])
```

```python
df.value_counts("Vallitseva sää")
```

```python
df1 = df[df["Vallitseva sää"].isin(["Ei merkittäviä sääilmiöitä", "Heikkoa lumisadetta", "Heikkoa vesisadetta", "Utua"])]
```

```python
df1.shape
```

```python
df1.to_csv("data/clf_imbalanced_weather.csv", index=False)
```

```python
df = df.replace("-", None)
df["Ilman lämpötila keskiarvo [°C]"]=df["Ilman lämpötila keskiarvo [°C]"].astype(float)
```

```python
print(df["Ilman lämpötila keskiarvo [°C]"].max())
```

```python
print(df["Ilman lämpötila keskiarvo [°C]"].min())
```

```python
df2 = df[(df["Ilman lämpötila keskiarvo [°C]"] <= -14) | ((df["Ilman lämpötila keskiarvo [°C]"] >= 0) & (df["Ilman lämpötila keskiarvo [°C]"] <= 2))]
```

```python
df2.shape
```

```python
df2.to_csv("data/reg_imbalanced_weather.csv", index=False)
```
