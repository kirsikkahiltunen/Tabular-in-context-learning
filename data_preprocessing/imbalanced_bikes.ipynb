```python
df = pd.read_csv("data/EN_Bikes.csv")
```

```python
df.value_counts("Return station name")
```

```python
df1 = df[df["Return station name"].isin(["Munkkiniemen aukio", "Innopoli", "Puotinkylän kartano", "Linnuntie", "Huovitie"])]
```

```python
df1.shape
```

```python
df1.value_counts("Return station name")
```

```python
df1 = df1.dropna()
```

```python
df1.shape
```

```python
df1.to_csv("data/clf_imbalanced_bikes.csv", index=False)
```

```python
df.value_counts("Covered distance (m)")
```

```python
df = df.replace("-", None)
df["Covered distance (m)"]=df["Covered distance (m)"].astype(float)
```

```python
df2 = df[((df["Covered distance (m)"] >= 0 ) & (df["Covered distance (m)"] <= 1100)) | (df["Covered distance (m)"] >= 7000)]
```

```python
df2.value_counts("Covered distance (m)")
```

```python
df2.shape
```

```python
df2.to_csv("data/reg_imbalanced_bikes.csv", index=False)
```
